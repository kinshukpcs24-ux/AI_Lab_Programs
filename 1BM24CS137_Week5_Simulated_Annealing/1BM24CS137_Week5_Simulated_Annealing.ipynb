```python
import random
import math

def get_conflicts(state):
    """Calculate the number of mutual attacks between queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict (since each queen is in a unique column, column conflicts are impossible)
            if state[i] == state[j]:
                conflicts += 1
            # Diagonal conflicts
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbor(state):
    """Generate a neighbor state by moving a random queen to a random different row."""
    neighbor = list(state)
    col = random.randint(0, len(state) - 1)
    row = random.randint(0, len(state) - 1)
    while row == state[col]:
        row = random.randint(0, len(state) - 1)
    neighbor[col] = row
    return neighbor

def simulated_annealing_n_queens(n=8, max_iter=10000, initial_temp=100.0, cooling_rate=0.995):
    """Solve N-Queens using Simulated Annealing."""
    current_state = [random.randint(0, n - 1) for _ in range(n)]
    current_cost = get_conflicts(current_state)

    temp = initial_temp

    for i in range(max_iter):
        if current_cost == 0:
            print(f"Solution found at iteration {i}!")
            return current_state, current_cost

        neighbor = get_neighbor(current_state)
        neighbor_cost = get_conflicts(neighbor)

        cost_diff = neighbor_cost - current_cost
        if cost_diff < 0:
            current_state = neighbor
            current_cost = neighbor_cost
        else:
            acceptance_prob = math.exp(-cost_diff / temp) if temp > 0 else 0
            if random.random() < acceptance_prob:
                current_state = neighbor
                current_cost = neighbor_cost

        # Cool down
        temp *= cooling_rate

    return current_state, current_cost

def print_board(state):
    """Utility to display the board."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

solution, final_conflicts = simulated_annealing_n_queens(n=8)
print(f"Final Board State (Row positions per column): {solution}")
print(f"Remaining Conflicts: {final_conflicts}\n")
print_board(solution)
```
