```python
import ast

def alpha_beta_search(state, tree):
    if not isinstance(tree, dict):
        return None

    actions = list(tree.keys())
    best_action = None
    v = float('-inf')
    alpha = float('-inf')
    beta = float('inf')

    for action in actions:
        child_state = tree[action]
        val = min_value(child_state, alpha, beta)
        if val > v:
            v = val
            best_action = action
        alpha = max(alpha, v)

    return best_action

def max_value(state, alpha, beta):
    if not isinstance(state, dict):
        return state

    v = float('-inf')
    for action in state:
        v = max(v, min_value(state[action], alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v

def min_value(state, alpha, beta):
    if not isinstance(state, dict):
        return state

    v = float('inf')
    for action in state:
        v = min(v, max_value(state[action], alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v

# Example of tree format: {"A": {"C": 3, "D": 5}, "B": {"E": 6, "F": 9}}
user_input = input("Enter your game tree as a dictionary (or press Enter to use the default game tree): ")

if user_input.strip() == "":
    game_tree = {
        'A': {
            'C': 3,
            'D': 5
        },
        'B': {
            'E': 6,
            'F': 9
        }
    }
else:
    try:
        game_tree = ast.literal_eval(user_input)
        if not isinstance(game_tree, dict):
            raise ValueError("Input must be a valid dictionary structure.")
    except Exception as e:
        print(f"Invalid input format ({e}). Using default game tree instead.")
        game_tree = {
            'A': {
                'C': 3,
                'D': 5
            },
            'B': {
                'E': 6,
                'F': 9
            }
        }

print("\nUsing Game Tree:", game_tree)
best_action = alpha_beta_search('Root', game_tree)
optimal_val = max_value(game_tree, float('-inf'), float('inf'))

print(f"Best Action from Root: {best_action}")
print(f"Optimal Value: {optimal_val}")
```
